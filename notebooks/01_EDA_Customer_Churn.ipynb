```python
# ============================================
# Telecom Services vs Churn Analysis
# ============================================

# Create a temporary copy so the original df is not modified
services_df = df.copy()


# ============================================
# Define Telecom Service Columns
# ============================================

service_columns = [
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "PhoneService"
]


# ============================================
# 1. Calculate Churn Percentage for Each
#    Telecom Service Category
# ============================================

print("Churn Percentage by Telecom Service Category:")

for column in service_columns:
    churn_percentage = (
        services_df.groupby(column)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .reset_index(name="Churn Percentage")
    )
    
    print(f"\n{'=' * 60}")
    print(f"{column}")
    print(f"{'=' * 60}")
    
    display(churn_percentage)


# ============================================
# 2. Visualize Churn Percentage for Each
#    Telecom Service
# ============================================

for column in service_columns:
    
    # Calculate churn percentage
    churn_percentage = (
        services_df.groupby(column)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .reset_index(name="Churn Percentage")
    )
    
    # Create individual plot
    plt.figure(figsize=(8, 5))
    
    ax = sns.barplot(
        data=churn_percentage,
        x=column,
        y="Churn Percentage"
    )
    
    plt.title(f"Churn Percentage by {column}")
    plt.xlabel(column)
    plt.ylabel("Churn Percentage (%)")
    plt.xticks(rotation=30)
    
    # Display percentage values on bars
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.1f%%",
            padding=3
        )
    
    plt.tight_layout()
    plt.show()


# ============================================
# 3. Create TotalServices Feature
# ============================================
# Count additional services subscribed by each customer.
# InternetService and PhoneService are considered services.
# The following add-on services count as 1 when the customer
# has subscribed and 0 otherwise:
# OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport,
# StreamingTV, StreamingMovies

additional_service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

# Create binary indicators temporarily and calculate total services
services_df["TotalServices"] = (
    (services_df["InternetService"] != "No").astype(int)
    + (services_df["PhoneService"] == "Yes").astype(int)
    + (services_df[additional_service_columns] == "Yes").sum(axis=1)
)


# ============================================
# 4. Display TotalServices Distribution
# ============================================

print("Distribution of TotalServices:")
display(services_df["TotalServices"].value_counts().sort_index())


# ============================================
# 5. Calculate Churn Percentage Based on
#    TotalServices
# ============================================

total_services_churn = (
    services_df.groupby("TotalServices")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Churn Percentage")
)

print("Churn Percentage by TotalServices:")
display(total_services_churn)


# ============================================
# 6. Visualize Churn Percentage Based on
#    TotalServices
# ============================================

plt.figure(figsize=(9, 6))

ax = sns.barplot(
    data=total_services_churn,
    x="TotalServices",
    y="Churn Percentage"
)

plt.title("Churn Percentage by Number of Telecom Services")
plt.xlabel("Total Number of Services")
plt.ylabel("Churn Percentage (%)")

# Display percentage values on bars
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3
    )

plt.tight_layout()
plt.show()
```

```python
# ============================================
# Demographic & Billing Features vs Churn
# ============================================

# Create a temporary copy so the original df is not modified
demographic_df = df.copy()


# ============================================
# Define Features to Analyze
# ============================================

categorical_features = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PaperlessBilling",
    "PaymentMethod"
]


# ============================================
# Interpret SeniorCitizen Values
# ============================================

# Create a temporary readable version only for analysis
demographic_df["SeniorCitizenLabel"] = demographic_df["SeniorCitizen"].map({
    0: "Non-Senior Citizen",
    1: "Senior Citizen"
})


# ============================================
# 1. Calculate Churn Percentage for Each Feature
# ============================================

print("Churn Percentage by Category:")

for feature in categorical_features:

    # Use readable labels for SeniorCitizen
    group_column = (
        "SeniorCitizenLabel"
        if feature == "SeniorCitizen"
        else feature
    )

    churn_percentage = (
        demographic_df.groupby(group_column)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .reset_index(name="Churn Percentage")
    )

    print(f"\n{'=' * 60}")
    print(f"{feature}")
    print(f"{'=' * 60}")

    display(churn_percentage)


# ============================================
# 2. Visualize Churn Percentage for Each Feature
# ============================================

for feature in categorical_features:

    # Use readable labels for SeniorCitizen
    group_column = (
        "SeniorCitizenLabel"
        if feature == "SeniorCitizen"
        else feature
    )

    # Calculate churn percentage
    churn_percentage = (
        demographic_df.groupby(group_column)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .reset_index(name="Churn Percentage")
    )

    # Create individual plot
    plt.figure(figsize=(9, 6))

    ax = sns.barplot(
        data=churn_percentage,
        x=group_column,
        y="Churn Percentage"
    )

    # Chart title and labels
    if feature == "SeniorCitizen":
        plt.title("Churn Percentage by Senior Citizen Status")
        plt.xlabel("Customer Type")
    else:
        plt.title(f"Churn Percentage by {feature}")
        plt.xlabel(feature)

    plt.ylabel("Churn Percentage (%)")
    plt.xticks(rotation=30)

    # Display percentage values on bars
    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.1f%%",
            padding=3
        )

    plt.tight_layout()
    plt.show()
```

```python
# ============================================
# Correlation Analysis of Numerical Variables
# ============================================

# Create a temporary copy so the original df is not modified
correlation_df = df.copy()


# ============================================
# 1. Select Numerical Columns
# ============================================

numerical_columns = correlation_df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical Columns:")
print(numerical_columns)


# ============================================
# 2. Convert Churn into Numerical Representation
#    Only for Correlation Analysis
# ============================================

correlation_df["Churn_Numeric"] = correlation_df["Churn"].map({
    "No": 0,
    "Yes": 1
})


# ============================================
# 3. Create Correlation Matrix
# ============================================

# Add numerical Churn target to numerical columns
correlation_columns = numerical_columns + ["Churn_Numeric"]

correlation_matrix = correlation_df[correlation_columns].corr()

print("Correlation Matrix:")
display(correlation_matrix)


# ============================================
# 4. Display Correlation Matrix as Heatmap
# ============================================

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Variables")
plt.xlabel("Variables")
plt.ylabel("Variables")

plt.tight_layout()
plt.show()


# ============================================
# 5. Correlation of Numerical Features with Churn
#    Sorted from Highest to Lowest
# ============================================

churn_correlation = (
    correlation_matrix["Churn_Numeric"]
    .drop("Churn_Numeric")
    .sort_values(ascending=False)
)

print("Correlation of Numerical Features with Churn:")
display(churn_correlation)
```

```python
# ============================================
# High-Risk Customer Segment Analysis
# ============================================

# Create a temporary copy so the original df is not modified
segment_df = df.copy()


# ============================================
# 1. Create Meaningful Tenure Groups
# ============================================

tenure_bins = [0, 12, 24, 48, float("inf")]
tenure_labels = [
    "0–12 months",
    "13–24 months",
    "25–48 months",
    "49+ months"
]

segment_df["TenureGroup"] = pd.cut(
    segment_df["tenure"],
    bins=tenure_bins,
    labels=tenure_labels,
    include_lowest=True
)


# ============================================
# 2. Define Features for Segment Analysis
# ============================================

segment_features = [
    "Contract",
    "InternetService",
    "TenureGroup",
    "PaymentMethod",
    "TechSupport",
    "OnlineSecurity"
]


# ============================================
# 3. Set Minimum Customer Count
# ============================================

# Ignore segments with fewer than 50 customers
# to avoid misleadingly high churn percentages
MIN_CUSTOMERS = 50


# ============================================
# 4. Create Grouped Aggregation
# ============================================

high_risk_segments = (
    segment_df
    .groupby(segment_features, observed=True)
    .agg(
        CustomerCount=("Churn", "size"),
        ChurnCount=("Churn", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)


# ============================================
# 5. Calculate Churn Percentage
# ============================================

high_risk_segments["ChurnPercentage"] = (
    high_risk_segments["ChurnCount"]
    / high_risk_segments["CustomerCount"]
    * 100
)


# ============================================
# 6. Filter Out Very Small Customer Segments
# ============================================

high_risk_segments = high_risk_segments[
    high_risk_segments["CustomerCount"] >= MIN_CUSTOMERS
]


# ============================================
# 7. Sort Segments by Churn Percentage
# ============================================

high_risk_segments = high_risk_segments.sort_values(
    by="ChurnPercentage",
    ascending=False
).reset_index(drop=True)


# ============================================
# 8. Display Top High-Risk Customer Segments
# ============================================

print("Top High-Risk Customer Segments:")
display(
    high_risk_segments.head(20)
)


# ============================================
# 9. Display Segments with Highest Churn
#    and Their Key Metrics
# ============================================

print("Top 10 High-Risk Segments:")

display(
    high_risk_segments[
        segment_features
        + [
            "CustomerCount",
            "ChurnCount",
            "ChurnPercentage"
        ]
    ].head(10)
)
```

```python
# ============================================
# Final EDA Summary
# ============================================

# Create a temporary copy so the original df is not modified
summary_df = df.copy()


# ============================================
# 1. Dataset Size
# ============================================

print("=" * 60)
print("FINAL EDA SUMMARY")
print("=" * 60)

print("\n1. Dataset Size:")
print(f"Number of Rows: {summary_df.shape[0]}")
print(f"Number of Columns: {summary_df.shape[1]}")


# ============================================
# 2. Overall Churn Rate
# ============================================

overall_churn_rate = (
    (summary_df["Churn"] == "Yes").mean() * 100
)

print("\n2. Overall Churn Rate:")
print(f"{overall_churn_rate:.2f}%")


# ============================================
# 3. Number of Churned and Non-Churned Customers
# ============================================

churn_counts = summary_df["Churn"].value_counts()

print("\n3. Customer Churn Counts:")
print(f"Churned Customers: {churn_counts.get('Yes', 0)}")
print(f"Non-Churned Customers: {churn_counts.get('No', 0)}")


# ============================================
# 4. Average Tenure by Churn Status
# ============================================

average_tenure = (
    summary_df.groupby("Churn")["tenure"]
    .mean()
    .reset_index(name="Average Tenure")
)

print("\n4. Average Tenure by Churn Status:")
display(average_tenure)


# ============================================
# 5. Average MonthlyCharges by Churn Status
# ============================================

average_monthly_charges = (
    summary_df.groupby("Churn")["MonthlyCharges"]
    .mean()
    .reset_index(name="Average MonthlyCharges")
)

print("\n5. Average MonthlyCharges by Churn Status:")
display(average_monthly_charges)


# ============================================
# 6. Churn Rate by Contract Type
# ============================================

contract_churn_rate = (
    summary_df.groupby("Contract")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Churn Rate (%)")
    .sort_values("Churn Rate (%)", ascending=False)
)

print("\n6. Churn Rate by Contract Type:")
display(contract_churn_rate)


# ============================================
# 7. Churn Rate by InternetService
# ============================================

internet_churn_rate = (
    summary_df.groupby("InternetService")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Churn Rate (%)")
    .sort_values("Churn Rate (%)", ascending=False)
)

print("\n7. Churn Rate by InternetService:")
display(internet_churn_rate)


# ============================================
# 8. Churn Rate by PaymentMethod
# ============================================

payment_churn_rate = (
    summary_df.groupby("PaymentMethod")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Churn Rate (%)")
    .sort_values("Churn Rate (%)", ascending=False)
)

print("\n8. Churn Rate by PaymentMethod:")
display(payment_churn_rate)


# ============================================
# 9. Top 5 High-Risk Customer Segments
# ============================================

# Create meaningful tenure groups
tenure_bins = [0, 12, 24, 48, float("inf")]
tenure_labels = [
    "0–12 months",
    "13–24 months",
    "25–48 months",
    "49+ months"
]

summary_df["TenureGroup"] = pd.cut(
    summary_df["tenure"],
    bins=tenure_bins,
    labels=tenure_labels,
    include_lowest=True
)


# Define segment features
segment_features = [
    "Contract",
    "InternetService",
    "TenureGroup",
    "PaymentMethod",
    "TechSupport",
    "OnlineSecurity"
]


# Minimum customer count to avoid unreliable small segments
MIN_CUSTOMERS = 50


# Calculate segment metrics
high_risk_segments = (
    summary_df
    .groupby(segment_features, observed=True)
    .agg(
        CustomerCount=("Churn", "size"),
        ChurnCount=("Churn", lambda x: (x == "Yes").sum())
    )
    .reset_index()
)


# Calculate churn percentage
high_risk_segments["ChurnPercentage"] = (
    high_risk_segments["ChurnCount"]
    / high_risk_segments["CustomerCount"]
    * 100
)


# Filter small segments
high_risk_segments = high_risk_segments[
    high_risk_segments["CustomerCount"] >= MIN_CUSTOMERS
]


# Sort by churn percentage
high_risk_segments = high_risk_segments.sort_values(
    by="ChurnPercentage",
    ascending=False
)


print("\n9. Top 5 High-Risk Customer Segments:")

display(
    high_risk_segments[
        segment_features
        + [
            "CustomerCount",
            "ChurnCount",
            "ChurnPercentage"
        ]
    ].head(5)
)
```
