```python
# ============================================
# 1. Import Required Libraries
# ============================================

import pandas as pd
import numpy as np


# ============================================
# 2. Load the Dataset
# ============================================

df = pd.read_csv("Telco-Customer-Churn.csv")


# ============================================
# 3. Display Dataset Shape
# ============================================

print("Shape of the Dataset:")
print(df.shape)


# ============================================
# 4. Display First 5 Rows
# ============================================

print("\nFirst 5 Rows:")
display(df.head())


# ============================================
# 5. Display Data Types
# ============================================

print("\nData Types Before Conversion:")
display(df.dtypes)


# ============================================
# 6. Convert TotalCharges to Numeric
# ============================================

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)


# ============================================
# 7. Display Missing Values After Conversion
# ============================================

print("\nMissing Values After TotalCharges Conversion:")
print(df["TotalCharges"].isnull().sum())


# ============================================
# 8. Display Unique Values of Churn
# ============================================

print("\nUnique Values in Churn Column:")
print(df["Churn"].unique())
```

```python
# ============================================
# 1. Remove CustomerID Column
# ============================================

df = df.drop(columns=["customerID"])


# ============================================
# 2. Convert Churn from Yes/No to 1/0
# ============================================

df["Churn"] = df["Churn"].map({
    "No": 0,
    "Yes": 1
})


# ============================================
# 3. Check Missing Values After TotalCharges
#    Conversion
# ============================================

print("Missing Values in TotalCharges:")
print(df["TotalCharges"].isnull().sum())


# ============================================
# 4. Handle Missing TotalCharges Values
# ============================================

# Fill missing TotalCharges values with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)


# ============================================
# 5. Display Shape After Operations
# ============================================

print("\nShape of DataFrame:")
print(df.shape)


# ============================================
# 6. Display First 5 Rows
# ============================================

print("\nFirst 5 Rows:")
display(df.head())


# ============================================
# 7. Display Data Types of All Columns
# ============================================

print("\nData Types of All Columns:")
display(df.dtypes)
```

```python
# ============================================
# Create TenureGroup Feature
# ============================================

# Define tenure intervals and corresponding labels
tenure_bins = [0, 12, 24, 48, float("inf")]

tenure_labels = [
    "New",
    "Growing",
    "Established",
    "Loyal"
]

# Create TenureGroup without modifying the original tenure column
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=tenure_bins,
    labels=tenure_labels,
    include_lowest=True
)


# ============================================
# Display Value Counts of Each TenureGroup
# ============================================

print("Customer Count by TenureGroup:")
display(
    df["TenureGroup"]
    .value_counts()
    .reindex(tenure_labels)
)


# ============================================
# Calculate Churn Rate for Each TenureGroup
# ============================================

tenure_churn_rate = (
    df.groupby("TenureGroup", observed=False)["Churn"]
    .agg(
        CustomerCount="count",
        ChurnedCustomers="sum",
        ChurnRate=lambda x: x.mean() * 100
    )
    .reindex(tenure_labels)
    .reset_index()
)


# ============================================
# Display Churn Rate Results
# ============================================

print("\nChurn Rate by TenureGroup:")
display(tenure_churn_rate)
```

```python
# ============================================
# Create TotalServices Feature
# ============================================

# Define additional service columns
service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

# Count only "Yes" as 1.
# "No" and "No internet service" are treated as 0.
df["TotalServices"] = (
    df[service_columns] == "Yes"
).sum(axis=1)


# ============================================
# Display Minimum, Maximum, and Average
# ============================================

print("TotalServices Statistics:")

print(f"Minimum TotalServices: {df['TotalServices'].min()}")
print(f"Maximum TotalServices: {df['TotalServices'].max()}")
print(f"Average TotalServices: {df['TotalServices'].mean():.2f}")


# ============================================
# Display Value Counts of TotalServices
# ============================================

print("\nValue Counts of TotalServices:")

display(
    df["TotalServices"]
    .value_counts()
    .sort_index()
)


# ============================================
# Calculate Churn Rate for Each TotalServices Value
# ============================================

total_services_churn_rate = (
    df.groupby("TotalServices")["Churn"]
    .agg(
        CustomerCount="count",
        ChurnedCustomers="sum",
        ChurnRate=lambda x: x.mean() * 100
    )
    .reset_index()
)


# ============================================
# Display Churn Rate Results
# ============================================

print("\nChurn Rate by TotalServices:")

display(total_services_churn_rate)
```

```python
# ============================================
# Create Business-Oriented Customer Value Features
# ============================================

# Create Estimated Annual Customer Value
# without modifying the original MonthlyCharges column
df["EstimatedAnnualValue"] = df["MonthlyCharges"] * 12

# Create Estimated Two-Year Customer Value
df["EstimatedTwoYearValue"] = df["MonthlyCharges"] * 24


# ============================================
# Create CustomerValueSegment Using qcut
# ============================================

df["CustomerValueSegment"] = pd.qcut(
    df["EstimatedAnnualValue"],
    q=3,
    labels=[
        "Low Value",
        "Medium Value",
        "High Value"
    ]
)


# ============================================
# Display Summary Statistics
# ============================================

print("Summary Statistics for Customer Value Features:")

display(
    df[
        [
            "MonthlyCharges",
            "EstimatedAnnualValue",
            "EstimatedTwoYearValue"
        ]
    ].describe()
)


# ============================================
# Display Number of Customers in Each
# CustomerValueSegment
# ============================================

print("Number of Customers in Each CustomerValueSegment:")

display(
    df["CustomerValueSegment"]
    .value_counts()
    .reindex([
        "Low Value",
        "Medium Value",
        "High Value"
    ])
)


# ============================================
# Calculate Churn Rate for Each
# CustomerValueSegment
# ============================================

customer_value_churn_rate = (
    df.groupby("CustomerValueSegment", observed=False)["Churn"]
    .agg(
        CustomerCount="count",
        ChurnedCustomers="sum",
        ChurnRate=lambda x: x.mean() * 100
    )
    .reindex([
        "Low Value",
        "Medium Value",
        "High Value"
    ])
    .reset_index()
)


# ============================================
# Display Churn Rate Results
# ============================================

print("Churn Rate by CustomerValueSegment:")

display(customer_value_churn_rate)
```

```python
# ============================================
# Create EstimatedLifetimeValue Feature
# ============================================

# Calculate estimated lifetime value
# without modifying the original MonthlyCharges or tenure columns
df["EstimatedLifetimeValue"] = (
    df["MonthlyCharges"] * df["tenure"]
)


# ============================================
# Create AverageMonthlyRevenue Feature
# ============================================

# Avoid division by zero for customers with tenure = 0
# by replacing zero tenure with NaN temporarily
df["AverageMonthlyRevenue"] = (
    df["TotalCharges"] /
    df["tenure"].replace(0, np.nan)
)


# ============================================
# Handle Invalid or Infinite Values
# ============================================

# Replace positive and negative infinite values with NaN
df["AverageMonthlyRevenue"] = df[
    "AverageMonthlyRevenue"
].replace([np.inf, -np.inf], np.nan)

# Fill invalid values with 0
# Customers with tenure = 0 are assigned 0 average monthly revenue
df["AverageMonthlyRevenue"] = df[
    "AverageMonthlyRevenue"
].fillna(0)


# ============================================
# Display Summary Statistics
# ============================================

print("Summary Statistics for New Features:")

display(
    df[
        [
            "EstimatedLifetimeValue",
            "AverageMonthlyRevenue"
        ]
    ].describe()
)


# ============================================
# Check for NaN Values
# ============================================

print("NaN Values in New Features:")

display(
    df[
        [
            "EstimatedLifetimeValue",
            "AverageMonthlyRevenue"
        ]
    ].isna().sum()
)


# ============================================
# Check for Infinite Values
# ============================================

print("Infinite Values in New Features:")

display(
    np.isinf(
        df[
            [
                "EstimatedLifetimeValue",
                "AverageMonthlyRevenue"
            ]
        ]
    ).sum()
)


# ============================================
# Compare Average EstimatedLifetimeValue
# by Churn Status
# ============================================

lifetime_value_by_churn = (
    df.groupby("Churn")["EstimatedLifetimeValue"]
    .mean()
    .reset_index(name="Average Estimated Lifetime Value")
)


# ============================================
# Display Results
# ============================================

print("Average EstimatedLifetimeValue by Churn Status:")

display(lifetime_value_by_churn)
```

```python
# ============================================
# Create Binary Customer Profile Features
# ============================================

# HasPartner: 1 if Partner is Yes, otherwise 0
df["HasPartner"] = (
    df["Partner"] == "Yes"
).astype(int)


# HasDependents: 1 if Dependents is Yes, otherwise 0
df["HasDependents"] = (
    df["Dependents"] == "Yes"
).astype(int)


# IsSeniorCitizen: Use existing SeniorCitizen
# column as a numerical binary feature
df["IsSeniorCitizen"] = df["SeniorCitizen"].astype(int)


# ============================================
# Display Value Counts
# ============================================

print("Value Counts for HasPartner:")
display(df["HasPartner"].value_counts().sort_index())

print("\nValue Counts for HasDependents:")
display(df["HasDependents"].value_counts().sort_index())

print("\nValue Counts for IsSeniorCitizen:")
display(df["IsSeniorCitizen"].value_counts().sort_index())


# ============================================
# Calculate Churn Rate for Each Category
# ============================================

print("Churn Rate by HasPartner:")
partner_churn_rate = (
    df.groupby("HasPartner")["Churn"]
    .agg(
        CustomerCount="count",
        ChurnedCustomers="sum",
        ChurnRate=lambda x: x.mean() * 100
    )
    .reset_index()
)

display(partner_churn_rate)


print("Churn Rate by HasDependents:")
dependents_churn_rate = (
    df.groupby("HasDependents")["Churn"]
    .agg(
        CustomerCount="count",
        ChurnedCustomers="sum",
        ChurnRate=lambda x: x.mean() * 100
    )
    .reset_index()
)

display(dependents_churn_rate)


print("Churn Rate by IsSeniorCitizen:")
senior_churn_rate = (
    df.groupby("IsSeniorCitizen")["Churn"]
    .agg(
        CustomerCount="count",
        ChurnedCustomers="sum",
        ChurnRate=lambda x: x.mean() * 100
    )
    .reset_index()
)

display(senior_churn_rate)


# ============================================
# Check Valid Binary Values
# ============================================

binary_features = [
    "HasPartner",
    "HasDependents",
    "IsSeniorCitizen"
]

print("Unique Values in Binary Features:")

for feature in binary_features:
    print(f"\n{feature}:")
    print(sorted(df[feature].unique()))


# ============================================
# Verify That Features Contain Only 0 and 1
# ============================================

print("\nBinary Feature Validation:")

for feature in binary_features:
    is_valid = df[feature].isin([0, 1]).all()
    print(f"{feature}: {'Valid (0 and 1 only)' if is_valid else 'Invalid values found'}")
```

```python
# ============================================
# Create Service-Related Binary Features
# ============================================

# HasTechSupport: 1 if TechSupport is Yes, otherwise 0
df["HasTechSupport"] = (
    df["TechSupport"] == "Yes"
).astype(int)


# HasOnlineSecurity: 1 if OnlineSecurity is Yes, otherwise 0
df["HasOnlineSecurity"] = (
    df["OnlineSecurity"] == "Yes"
).astype(int)


# HasOnlineBackup: 1 if OnlineBackup is Yes, otherwise 0
df["HasOnlineBackup"] = (
    df["OnlineBackup"] == "Yes"
).astype(int)


# HasDeviceProtection: 1 if DeviceProtection is Yes, otherwise 0
df["HasDeviceProtection"] = (
    df["DeviceProtection"] == "Yes"
).astype(int)


# ============================================
# Display Value Counts
# ============================================

service_features = [
    "HasTechSupport",
    "HasOnlineSecurity",
    "HasOnlineBackup",
    "HasDeviceProtection"
]

print("Value Counts for Service-Related Features:")

for feature in service_features:
    print(f"\n{feature}:")
    display(
        df[feature]
        .value_counts()
        .sort_index()
    )


# ============================================
# Calculate Churn Rate for Each Feature
# ============================================

print("Churn Rate by Service-Related Features:")

for feature in service_features:
    
    churn_rate = (
        df.groupby(feature)["Churn"]
        .agg(
            CustomerCount="count",
            ChurnedCustomers="sum",
            ChurnRate=lambda x: x.mean() * 100
        )
        .reset_index()
    )
    
    print(f"\n{'=' * 50}")
    print(f"Churn Rate by {feature}")
    print(f"{'=' * 50}")
    
    display(churn_rate)
```

```python
# ============================================
# Create HighMonthlyCharges Feature
# ============================================

# NOTE:
# For this feature engineering demonstration, we calculate
# the median from the complete df because train-test split
# has not been performed yet.
#
# In the final production ML pipeline, the median threshold
# MUST be calculated only from the training data to avoid
# data leakage. The same training median should then be
# applied to validation, test, and new customer data.

# ============================================
# 1. Calculate Median MonthlyCharges
# ============================================

monthly_charges_median = df["MonthlyCharges"].median()

print("Median MonthlyCharges Threshold:")
print(monthly_charges_median)


# ============================================
# 2. Create HighMonthlyCharges Feature
# ============================================

# 1 = MonthlyCharges greater than the median
# 0 = MonthlyCharges less than or equal to the median
df["HighMonthlyCharges"] = (
    df["MonthlyCharges"] > monthly_charges_median
).astype(int)


# ============================================
# 3. Display the Threshold
# ============================================

print("\nThreshold Used for HighMonthlyCharges:")
print(f"MonthlyCharges Median = {monthly_charges_median:.2f}")


# ============================================
# 4. Display Value Counts
# ============================================

print("\nValue Counts of HighMonthlyCharges:")

display(
    df["HighMonthlyCharges"]
    .value_counts()
    .sort_index()
)


# ============================================
# 5. Calculate Churn Rate for
#    HighMonthlyCharges = 0 and 1
# ============================================

high_monthly_charges_churn_rate = (
    df.groupby("HighMonthlyCharges")["Churn"]
    .agg(
        CustomerCount="count",
        ChurnedCustomers="sum",
        ChurnRate=lambda x: x.mean() * 100
    )
    .reset_index()
)


# ============================================
# Display Churn Rate Results
# ============================================

print("\nChurn Rate by HighMonthlyCharges:")

display(high_monthly_charges_churn_rate)
```

```python
# ============================================
# Final Feature Engineering Review
# ============================================


# ============================================
# 1. Display Shape of the Dataset
# ============================================

print("=" * 60)
print("1. DATASET SHAPE")
print("=" * 60)

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")


# ============================================
# 2. Display All Column Names
# ============================================

print("\n" + "=" * 60)
print("2. ALL COLUMN NAMES")
print("=" * 60)

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")


# ============================================
# 3. Display First 5 Rows
# ============================================

print("\n" + "=" * 60)
print("3. FIRST 5 ROWS")
print("=" * 60)

display(df.head())


# ============================================
# 4. Display Data Types
# ============================================

print("\n" + "=" * 60)
print("4. DATA TYPES")
print("=" * 60)

display(df.dtypes)


# ============================================
# 5. Display Missing Values
# ============================================

print("\n" + "=" * 60)
print("5. MISSING VALUES")
print("=" * 60)

missing_values = df.isnull().sum()

display(
    missing_values[missing_values > 0]
)

if missing_values.sum() == 0:
    print("No missing values found in the dataset.")


# ============================================
# 6. Display Duplicate Rows
# ============================================

print("\n" + "=" * 60)
print("6. DUPLICATE ROWS")
print("=" * 60)

duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")


# ============================================
# 7. Display Summary Statistics
# ============================================

print("\n" + "=" * 60)
print("7. SUMMARY STATISTICS")
print("=" * 60)

display(
    df.describe(include="all").T
)


# ============================================
# 8. Display Newly Engineered Features
# ============================================

print("\n" + "=" * 60)
print("8. NEWLY ENGINEERED FEATURES")
print("=" * 60)

engineered_features = [
    "TenureGroup",
    "TotalServices",
    "EstimatedAnnualValue",
    "EstimatedTwoYearValue",
    "CustomerValueSegment",
    "EstimatedLifetimeValue",
    "AverageMonthlyRevenue",
    "HasPartner",
    "HasDependents",
    "IsSeniorCitizen",
    "HasTechSupport",
    "HasOnlineSecurity",
    "HasOnlineBackup",
    "HasDeviceProtection",
    "HighMonthlyCharges"
]

# Display only engineered features that currently exist in df
existing_engineered_features = [
    feature
    for feature in engineered_features
    if feature in df.columns
]

print("Engineered Features:")
for feature in existing_engineered_features:
    print(f"- {feature}")


# ============================================
# 9. Check for Infinite Values
# ============================================

print("\n" + "=" * 60)
print("9. INFINITE VALUES CHECK")
print("=" * 60)

# Select only numerical columns
numerical_df = df.select_dtypes(include=np.number)

infinite_values = np.isinf(numerical_df).sum()

# Display columns containing infinite values
infinite_columns = infinite_values[
    infinite_values > 0
]

if len(infinite_columns) == 0:
    print("No infinite values found in numerical columns.")
else:
    print("Infinite values found:")
    display(infinite_columns)


# ============================================
# 10. Concise Final Feature Engineering Summary
# ============================================

print("\n" + "=" * 60)
print("10. FINAL FEATURE ENGINEERING SUMMARY")
print("=" * 60)

print(f"Final dataset shape: {df.shape}")
print(f"Total features: {df.shape[1]}")
print(f"Total engineered features added: {len(existing_engineered_features)}")
print(f"Missing values: {df.isnull().sum().sum()}")
print(f"Duplicate rows: {df.duplicated().sum()}")

if len(infinite_columns) == 0:
    print("Infinite values: None")
else:
    print(f"Infinite values: Found in {len(infinite_columns)} column(s)")

print("\nFeature engineering review completed.")
print("Next steps: Encoding → Train-Test Split → Scaling (if required) → Class Imbalance Handling → Model Training")
```
